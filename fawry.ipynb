Data cleaning

```python
# 1. استخراج الأرقام والنقاط العشرية فقط والتخلص من أي حروف أو كلمات
fawry["Amount"] = fawry["Amount"].astype(str).str.extract(r"(\d+\.?\d*)")[0]

# 2. تحويل العمود لنوع رقمي float
fawry["Amount"] = pd.to_numeric(fawry["Amount"], errors="coerce")

```

```python
#ملينا ال null value بال  mean
fawry["Amount"]=fawry["Amount"].fillna(fawry["Amount"].mean())
fawry["Fawry_Fee"]=fawry["Fawry_Fee"].fillna(fawry["Fawry_Fee"].mean())
```

```python
#حولنا التاريخ من object to date
fawry['Transaction_Date'] = pd.to_datetime(fawry['Transaction_Date'], format='mixed')
```

```python
#حولنا اي تاريخ مستقبلي الى تاريخ اليوم
#today = pd.Timestamp.now()
#fawry.loc[fawry["Transaction_Date"] > today, "Transaction_Date"] = today
```

```python
print(fawry["Status"].unique())
fawry["Status"] = fawry["Status"].replace({
    "Panding": "Pending",
    "Succes": "Success"
})
```

```python
print(fawry["Payment_Method"].unique())
fawry["Payment_Method"] = fawry["Payment_Method"].replace({
    " Crdit Card ": "Credit Card"
})
```

```python
fawry["Merchant_ID"] = fawry["Merchant_ID"].astype(int)
```

Feature Engineering

```python
#اضافة عمود لكلا من (السنةورقم الشهرواسم الشهرورقم اليوم و اسم اليوم و وقت العملية و ساعة العملية)
fawry["Year"] = fawry["Transaction_Date"].dt.year
fawry["Month"] = fawry["Transaction_Date"].dt.month
fawry["Month_Name"] = fawry["Transaction_Date"].dt.month_name()
fawry["Day"] = fawry["Transaction_Date"].dt.day
fawry["Day_Name"] = fawry["Transaction_Date"].dt.day_name()
fawry["Time"] = fawry["Transaction_Date"].dt.time
fawry["Hour"] = fawry["Transaction_Date"].dt.hour
```

```python
fawry
```

```python
#اضافة عمود لحساب المبلغ الاجمالي
fawry["Total_Price"] = round(fawry["Fawry_Fee"]+ fawry["Amount"],2)
```

```python
#اضافة عمود لحساب ضريبة الشركة
fawry["Fee_Rate"] = round(fawry["Fawry_Fee"] / fawry["Total_Price"]*100,2)
```

```python
#دالة لعمل تصنيف بناءا علي السعر الاجمالي
def cat (x):
    if x < 100:
        return 'low'
    elif x < 1000:
        return 'Medium'
    elif x <  5000:
        return 'High'
    else :
        return 'Very High'
#اضافة عمود لتصنيف السعر
fawry["Price_Category"] = fawry["Total_Price"].apply(cat)
```

```python
#دالة لمعرفة مدي سرعة العملية
def speed (y):
    if y < 5:
        return 'Fast'
    elif y < 10:
        return 'Normal'
    else:
        return'Slow'
#اضافة عمود لمعرفة سرعة العملية
fawry["Processing_Speed"] = fawry["Processing_Time_Sec"].apply(speed)
```

```python
#عمود لتسهيل معرفة حالة العملية
fawry["Success_Status"] = fawry["Status"].map({"Success": 1,
                                               "Failed": 0,
                                               "Pending": -1}).astype(int)
```

RFM

```python
tdy = pd.Timestamp.today()
RFM = fawry.groupby("Merchant_ID").agg({
    "Transaction_Date" : lambda x:(tdy - x.max()).days,
    "Transaction_ID":"count",
    "Total_Price":"sum"
})
RFM.columns=["Recency","Frequency","Monetary"]
RFM["Monetary"] = RFM["Monetary"].round(2)
RFM
```

```python
#RFM["R-score"]=pd.qcut(RFM["Recency"].rank (method="first"),q=5,labels =[5,4,3,2,1])
RFM["R-score"]=pd.qcut(RFM["Recency"],q=5,labels =[5,4,3,2,1])

RFM["F-score"]=pd.qcut(RFM["Frequency"].rank (method="first"),q=5,labels =[1,2,3,4,5])

RFM["M-score"]=pd.qcut(RFM["Monetary"].rank (method="first"),q=5,labels = [1,2,3,4,5])

RFM
```
